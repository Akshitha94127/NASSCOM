```python
# 2A. DOT & CROSS PRODUCTS
import numpy as np
a = np.array([1, 2, 3])
b = np.array([4, 0, 1])
print('a . b  (dot)   :', np.dot(a, b))
print('a x b  (cross) :', np.cross(a, b))
```

```python
# 2B. NORMS (vector length) + COSINE SIMILARITY
import numpy as np
print('L2 norm  ||a||_2 :', la.norm(a))
print('L1 norm  ||a||_1 :', la.norm(a, 1))
print('Linf norm        :', la.norm(a, np.inf))
def cosine(u, v):
    return np.dot(u, v) / (la.norm(u) * la.norm(v))
print('cosine(a, b)     :', round(cosine(a, b), 3))
```

```python
# 3A. CORE MATRIX OPERATIONS
A = np.array([[2., 1.],
              [1., 3.]])
B = np.array([[1., 0.],
              [4., 2.]])
print('A @ B (matrix multiply):\n', A @ B)
print('A.T   (transpose):\n', A.T)
print('inverse(A):\n', la.inv(A))
print('trace(A) = sum of diagonal:', np.trace(A))
```

```python
# 3B. SPECIAL MATRICES: identity, symmetric, orthogonal
I = np.eye(2)
print('Identity I:\n', I)
print('A @ inv(A) == I ?', np.allclose(A @ la.inv(A), I))
print('A symmetric?', np.allclose(A, A.T))
theta = np.radians(30)
Q = np.array([[np.cos(theta), -np.sin(theta)],
              [np.sin(theta),  np.cos(theta)]])
print('Q orthogonal?', np.allclose(Q.T @ Q, I))
```

```python
# LAB EXERCISE 3 — Inverse, trace, and matrix tests
import numpy as np
M = np.array([[4., 2.], [2., 3.]])
P = np.array([[0., -1.], [1., 0.]])
# 1.
inv_M = np.linalg.inv(M)
trace_M = np.trace(M)
print("Inverse of M:\n", inv_M)
print("Trace of M:", trace_M)
# 2.
is_identity = np.allclose(M @ inv_M, np.eye(2))
print("M @ inverse(M) is Identity:", is_identity)
# 3.
is_symmetric = np.allclose(M, M.T)
is_orthogonal = np.allclose(P @ P.T, np.eye(2))
print("Is M symmetric?", is_symmetric)
print("Is P orthogonal?", is_orthogonal)
```

```python
# 4A. BUILD TRANSFORMATION MATRICES
import numpy as np
square = np.array([[0, 1, 1, 0],
                   [0, 0, 1, 1]], dtype=float)
S = np.array([[1.5, 0.0],
              [0.0, 0.5]])
t = np.radians(30)
R = np.array([[np.cos(t), -np.sin(t)],
              [np.sin(t),  np.cos(t)]])
scaled  = S @ square
rotated = R @ square
print('Rotated corners:\n', rotated)
```

```python
# 4B. PLOT ORIGINAL vs TRANSFORMED
def close_loop(pts):
    return np.hstack([pts, pts[:, :1]])
fig, ax = plt.subplots(figsize=(5, 5))
ax.plot(*close_loop(square),  marker='o', label='original')
ax.plot(*close_loop(scaled),  marker='o', label='scaled')
ax.plot(*close_loop(rotated), marker='o', label='rotated 30 deg')
ax.axhline(0, color='gray', lw=0.5); ax.axvline(0, color='gray', lw=0.5)
ax.set_aspect('equal'); ax.legend(); ax.set_title('Linear transformations')
plt.show()
```

```python
# LAB EXERCISE 4 — Apply 2 transformations & plot
import numpy as np
import matplotlib.pyplot as plt
tri = np.array([[0, 1, 0.5], [0, 0, 1.0]])
def close_loop(arr):
    return np.hstack([arr, arr[:, [0]]])
# 1.
S = np.array([
    [2.0, 0.0],
    [0.0, 0.5]
])
theta = np.radians(45)
R = np.array([
    [np.cos(theta), -np.sin(theta)],
    [np.sin(theta),  np.cos(theta)]
])
transformed_tri = R @ (S @ tri)
plt.figure(figsize=(6, 6))
tri_closed = close_loop(tri)
trans_closed = close_loop(transformed_tri)
plt.plot(tri_closed[0, :], tri_closed[1, :], 'b-o', label='Original')
plt.plot(trans_closed[0, :], trans_closed[1, :], 'r-s', label='Transformed (Scaled & Rotated)')
plt.axhline(0, color='gray', linestyle='--', linewidth=0.5)
plt.axvline(0, color='gray', linestyle='--', linewidth=0.5)
plt.grid(True)
plt.axis('equal')
plt.legend()
plt.title("Triangle Geometric Transformations")
plt.show()
```

```python
# 5A. COMPUTE EIGENVALUES & EIGENVECTORS
import numpy as np
A = np.array([[2., 0.],
              [0., 3.]])
vals, vecs = la.eig(A)
print('Eigenvalues  (lambda):', vals)
print('Eigenvectors (columns):\n', vecs)
```

```python
# 5B. VERIFY  A v = lambda v
for i in range(len(vals)):
    v = vecs[:, i]
    lhs = A @ v
    rhs = vals[i] * v
    print(f'lambda={vals[i]:.1f}  A v == lambda v ?', np.allclose(lhs, rhs))
```

```python
# LAB EXERCISE 5 — Eigenvalues/vectors & verification
import numpy as np
import scipy.linalg as la
C = np.array([[4., 1.],
              [2., 3.]])
# 1.
eigenvalues, eigenvectors = la.eig(C)
# 2.
print("Eigenvalues:", eigenvalues)
# 3.
lambda_1 = eigenvalues[0]
v_1 = eigenvectors[:, 0]
left_side = C @ v_1
right_side = lambda_1 * v_1
verification = np.allclose(left_side, right_side)
print("C @ v_1 == lambda_1 * v_1:", verification)
```

```python
# 6A. SOLVE A 3x3 SYSTEM  A x = b
import numpy as np
import numpy.linalg as la
A = np.array([[ 2.,  1., -1.],
              [-3., -1.,  2.],
              [-2.,  1.,  2.]])
b = np.array([8., -11., -3.])
x = la.solve(A, b)
print('Solution x:', x)
print('Rank of A :', la.matrix_rank(A))
print('Full rank -> a unique solution exists')
```
