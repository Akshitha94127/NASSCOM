```python
import numpy as np
import sympy as sp
x, y = sp.symbols('x y')
sp.init_printing()
np.random.seed(42)
print('Setup complete. SymPy', sp.__version__, '| NumPy', np.__version__)
```

```python
# 1A. NUMERICAL DERIVATIVE (finite difference)
def f(x):
    return x**2
h = 1e-6
slope_at_3 = (f(3 + h) - f(3)) / h
print('Numerical f\'(3):', round(slope_at_3, 4), ' (exact = 6)')
```

```python
# 1B. SYMBOLIC DERIVATIVE (SymPy)
expr = x**2
deriv = sp.diff(expr, x)
print('d/dx (x**2) =', deriv)
print('Symbolic f\'(3) =', deriv.subs(x, 3))
```

```python
# LAB EXERCISE 1 — Numerical vs symbolic derivative
import sympy as sp
def g(x):
    return x**3 + 2*x
# 1.
x_val = 2
h = 1e-6
numerical_deriv = (g(x_val + h) - g(x_val)) / h
print(f"Numerical Derivative: {numerical_deriv}")
# 2.
x = sp.Symbol('x')
symbolic_expr = g(x)
symbolic_deriv = sp.diff(symbolic_expr, x)
print(f"Symbolic Derivative Formula: {symbolic_deriv}")
# 3.
evaluated_symbolic = float(symbolic_deriv.subs(x, x_val))
print(f"Evaluated Symbolic Derivative: {evaluated_symbolic}")
print(f"Values Match? {abs(numerical_deriv - evaluated_symbolic) < 1e-5}")
```

```python
# 2A. PARTIAL DERIVATIVES
f2 = x**2 + 3*x*y + y**2
print('df/dx =', sp.diff(f2, x))
print('df/dy =', sp.diff(f2, y))
```

```python
# 2B. THE GRADIENT (vector of partials)
grad = [sp.diff(f2, v) for v in (x, y)]
print('grad f =', grad)
grad_at = [g.subs({x: 1, y: 2}) for g in grad]
print('grad f at (1, 2) =', grad_at)
```

```python
# LAB EXERCISE 2 — Compute a gradient
import sympy as sp
x, y = sp.symbols('x y')
h2 = x**2 * y + sp.sin(y)
dh_dx = sp.diff(h2, x)
dh_dy = sp.diff(h2, y)
gradient = [dh_dx, dh_dy]
print(f"Symbolic Gradient: {gradient}")
# 3.
evaluated_gradient = [expr.subs({x: 2, y: 0}) for expr in gradient]
print(f"Evaluated Gradient: {evaluated_gradient}")
```

```python
# 3A. CHAIN RULE BY HAND vs SymPy
by_hand = sp.cos(x**2) * 2*x
by_sympy = sp.diff(sp.sin(x**2), x)
print('By hand :', by_hand)
print('By SymPy:', by_sympy)
print('Match?  ', sp.simplify(by_hand - by_sympy) == 0)
```

```python
# 3B. CHAINING THREE FUNCTIONS
expr3 = (3*x + 1)**4
print('d/dx (3x+1)^4 =', sp.diff(expr3, x))
```

```python
# LAB EXERCISE 3 — Chain rule
import sympy as sp
x = sp.symbols('x')
by_hand = (sp.exp(x**2 + 1)) * (2 * x)
# 2.
y = sp.exp(x**2 + 1)
by_sympy = sp.diff(y, x)
print(f"By Hand Expression: {by_hand}")
print(f"By SymPy Expression: {by_sympy}")
# 3.
matches = sp.simplify(by_hand - by_sympy) == 0
print(f"Do the expressions match? {matches}")
```

```python
# 4A. FORWARD PASS
X = np.random.randn(4, 3)
Y = np.random.randn(4, 1)
W1 = np.random.randn(3, 5) * 0.1
W2 = np.random.randn(5, 1) * 0.1
z1 = X @ W1
h  = np.maximum(0, z1)
y_hat = h @ W2
loss = ((y_hat - Y) ** 2).mean()
print('Initial loss:', round(loss, 4))
```

```python
# 4B. BACKWARD PASS (gradients via the chain rule)
dy   = 2 * (y_hat - Y) / Y.size
dW2  = h.T @ dy
dh   = dy @ W2.T
dz1  = dh * (z1 > 0)
dW1  = X.T @ dz1
print('dW1 shape:', dW1.shape, '(matches W1)')
print('dW2 shape:', dW2.shape, '(matches W2)')
```

```python
# 4C. ONE GRADIENT-DESCENT STEP SHOULD LOWER THE LOSS
lr = 0.1
W1 -= lr * dW1
W2 -= lr * dW2
h_new = np.maximum(0, X @ W1)
loss_new = ((h_new @ W2 - Y) ** 2).mean()
print('Loss before:', round(loss, 4))
print('Loss after :', round(loss_new, 4), '-> should be lower')
```

```python
# LAB EXERCISE 4 — Derive the gradients for a 2-layer network
import numpy as np
Xb = np.random.randn(6, 4)
Yb = np.random.randn(6, 1)
Wa = np.random.randn(4, 8) * 0.1
Wb = np.random.randn(8, 1) * 0.1
# 1.
z1 = Xb @ Wa
h = np.maximum(0, z1)
y_hat = h @ Wb
loss_before = np.mean((y_hat - Yb) ** 2)
# 2.
m = Xb.shape[0]
dy = (2.0 / m) * (y_hat - Yb)
dWb = h.T @ dy
dh = dy @ Wb.T
dz1 = dh * (z1 > 0)
dWa = Xb.T @ dz1
# 3.
lr = 0.05
Wa -= lr * dWa
Wb -= lr * dWb
z1_new = Xb @ Wa
h_new = np.maximum(0, z1_new)
y_hat_new = h_new @ Wb
loss_after = np.mean((y_hat_new - Yb) ** 2)
print(f"Loss before gradient step: {loss_before:.6f}")
print(f"Loss after gradient step:  {loss_after:.6f}")
```

```python
# 5A. THE HESSIAN (matrix of second derivatives = curvature)
import sympy as sp
x, y = sp.symbols('x y')
f5 = x**2 + 3*x*y + y**2
H = sp.hessian(f5, (x, y))
print('Hessian of f:')
sp.pprint(H)
```

```python
# 5B. GRADIENT DESCENT ON A SIMPLE FUNCTION
import sympy as sp
x, y = sp.symbols('x y')
f = x**4 + y**2
hessian_mat = sp.hessian(f, (x, y))
print(f"Hessian Matrix:\n{hessian_mat}\n")
xv = 0.0
lr = 0.1
for step in range(20):
    grad = 2 * (xv - 7)
    xv = xv - lr * grad
print(f"Final x: {xv:.6f}")
```

```python
# LAB EXERCISE 5 — Hessian + gradient descent
import sympy as sp
# 1.
x, y = sp.symbols('x y')
f = x**4 + y**2
hessian_mat = sp.hessian(f, (x, y))
print(f"Hessian Matrix:\n{hessian_mat}\n")
# 2.
xv = 0.0
lr = 0.1
for step in range(20):
    grad = 2 * (xv - 7)
    xv = xv - lr * grad
print(f"Final x: {xv:.6f}")
```
