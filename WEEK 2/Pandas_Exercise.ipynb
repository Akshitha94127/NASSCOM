```python
import numpy as np
import pandas as pd
# Task 2: Synthetic Data Generation for Practice
print("\n--- Task 2: Synthetic Data Generation for Practice ---")
np.random.seed(42)
available_dates = pd.date_range(start="2023-01-01", end="2023-04-09", freq="D")
sampled_dates = np.random.choice(available_dates, size=100)
df_sales = pd.DataFrame(
    {
        "Date": sampled_dates,
        "Region": np.random.choice(["East", "West", "North", "South"], size=100),
        "Product": np.random.choice(
            ["Laptop", "Mouse", "Keyboard", "Monitor"], size=100
        ),
        "Sales": np.random.randint(100, 1001, size=100).astype(
            float
        ),
        "Quantity": np.random.randint(1, 11, size=100).astype(float),
    }
)
sales_nan_indices = np.random.choice(100, size=5, replace=False)
quantity_nan_indices = np.random.choice(100, size=3, replace=False)
df_sales.loc[sales_nan_indices, "Sales"] = np.nan
df_sales.loc[quantity_nan_indices, "Quantity"] = np.nan
print("\n--- First 10 Rows of df_sales ---")
print(df_sales.head(10))
print("\n--- Missing Value Count Verification ---")
print(df_sales.isna().sum())
```

```python
# Task 3: Indexing and Slicing in DataFrames
print("\n--- Task 3: Indexing and Slicing in DataFrames ---")
#task 3.1
df3_1 = df_sales[['Region', 'Sales']]
print("\n--- 3.1: df3_1 (First 5 Rows) ---")
print(df3_1.head())
#task 3.2
df3_2 = df_sales[df_sales['Product'] == 'Laptop']
print("\n--- 3.2: df3_2 (First 5 Rows) ---")
print(df3_2.head())
#task 3.3
df3_3 = df_sales[(df_sales['Region'] == 'East') & (df_sales['Sales'] > 500)]
print("\n--- 3.3: df3_3 (First 5 Rows) ---")
print(df3_3.head())
#task 3.4
monitor_df = df_sales[df_sales['Product'] == 'Monitor']
val3_4_iloc = monitor_df.iloc[0, monitor_df.columns.get_loc('Sales')]
first_monitor_label = monitor_df.index[0]
val3_4 = monitor_df.loc[first_monitor_label, 'Sales']
print("\n--- 3.4: val3_4 Value ---")
print(f"Sales value (via .iloc): {val3_4_iloc}")
print(f"Sales value (via .loc):  {val3_4}")
```

```python
# Task 4: Data Cleaning (Handling Missing Values)
print("\n--- Task 4: Data Cleaning (Handling Missing Values) ---")
#task 4.1
missing_counts4_1 = df_sales.isna().sum()
print("\n--- 4.1: missing_counts4_1 ---")
print(missing_counts4_1)
#task 4.2
df4_2_filled = df_sales.copy()
sales_mean = df4_2_filled['Sales'].mean()
df4_2_filled['Sales'] = df4_2_filled['Sales'].fillna(sales_mean)
print("\n--- 4.2: df4_2_filled 'Sales' Missing Count ---")
print(f"Remaining missing values in Sales: {df4_2_filled['Sales'].isna().sum()}")
print(f"Filled with Mean Value: {sales_mean:.2f}")
#task 4.3
df4_3_dropped = df_sales.dropna()
print("\n--- 4.3: df4_3_dropped Summary ---")
print(f"Original Row Count: {len(df_sales)}")
print(f"Cleaned Row Count:  {len(df4_3_dropped)}")
print(f"Total Rows Dropped: {len(df_sales) - len(df4_3_dropped)}")
```

```python
# Task 5: Data Manipulation (Columns, Sorting, Grouping)
print("\n--- Task 5: Data Manipulation (Columns, Sorting, Grouping) ---")
#task 5.1
df_sales['Total_Revenue'] = df_sales['Sales'] * df_sales['Quantity']
print("\n--- 5.1: df_sales with Total_Revenue (First 5 Rows) ---")
print(df_sales[['Sales', 'Quantity', 'Total_Revenue']].head())
#task 5.2
df5_2_sorted = df_sales.sort_values(by=['Date', 'Sales'], ascending=[True, False])
print("\n--- 5.2: df5_2_sorted (First 5 Rows) ---")
print(df5_2_sorted.head())
#task 5.3
df5_3_grouped_region = df_sales.groupby('Region')[['Sales', 'Quantity']].sum()
print("\n--- 5.3: df5_3_grouped_region ---")
print(df5_3_grouped_region)
#task 5.4
df5_4_grouped_product = df_sales.groupby('Product')[['Sales']].mean()
print("\n--- 5.4: df5_4_grouped_product ---")
print(df5_4_grouped_product)
```
